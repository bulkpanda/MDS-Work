```python
import json

df_fr = pd.read_excel("fetched_rows_2025.xlsx")

# Anon student columns
df_fr['student_number'] = df_fr['student_number'].apply(
    lambda x: anon_mapping.get(str(int(x)), x) if pd.notna(x) else x
)
df_fr['student_name'] = df_fr['student_number']  # replace name with anon ID
df_fr['student_email'] = df_fr['student_number'].apply(
    lambda x: f"{x}@anon.edu" if pd.notna(x) else x
)

# Build mappings from forms JSON for assessor names and patient DRNs
all_assessors = set()
all_drns = set()
for forms_str in df_fr['forms'].dropna():
    forms = json.loads(forms_str) if isinstance(forms_str, str) else forms_str
    for f in forms:
        if 'assessor_name' in f and f['assessor_name']:
            all_assessors.add(f['assessor_name'])
        if 'patient_drn' in f and f['patient_drn']:
            all_drns.add(str(f['patient_drn']))

assessor_mapping = {name: f"ASSESSOR_{i+1:03d}" for i, name in enumerate(sorted(all_assessors))}
drn_mapping = {drn: f"PT_{i+1:04d}" for i, drn in enumerate(sorted(all_drns))}

# Anonymize inside forms JSON
def anon_forms(forms_str):
    if pd.isna(forms_str):
        return forms_str
    forms = json.loads(forms_str) if isinstance(forms_str, str) else forms_str
    for f in forms:
        if 'assessor_name' in f:
            f['assessor_name'] = assessor_mapping.get(f['assessor_name'], f['assessor_name'])
        if 'patient_drn' in f:
            f['patient_drn'] = drn_mapping.get(str(f['patient_drn']), f['patient_drn'])
    return json.dumps(forms)

df_fr['forms'] = df_fr['forms'].apply(anon_forms)
```
